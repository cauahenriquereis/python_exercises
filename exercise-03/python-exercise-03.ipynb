Questão 1

```python
import numpy as np
import random

arr = np.ones(8)
arr2 = np.random.randint(0,9,8)

arrsoma = arr + arr2

if arrsoma.sum() >= 40:
  mat = arrsoma.reshape(4,2)
else:
  mat = arrsoma.reshape(2,4)

print(mat)
```

```python
mat = np.zeros([2,2])

linhaRandom = random.randint(0,1)
colunaRandom = random.randint(0,1)

mat[linhaRandom,colunaRandom] = 1

ganhou = True

for i in range(3):
  jogadaLinha = int(input("Digite a linha: "))
  jogadaColuna = int(input("Digite a coluna: "))

  if mat[jogadaLinha, jogadaColuna] == 1:
    print("Game Over! :( Try Again!")
    ganhou = False
    break

if ganhou:
  print("Congratulations! You beat the game! :)1")



```

```python
linhaRandom = random.randint(1,5)
colunaRandom = random.randint(1,5)

mat = np.empty([linhaRandom, colunaRandom])

linhas = mat.shape[0]
colunas = mat.shape[1]

mult = linhas * colunas

if mult % 2 == 0:
  print('Pode se tornar uma matriz com número par de elementos')
else:
  print ('Pode se tornar uma matriz com número ímpar de elementos')
```

Questão 4

```python
np.random.seed(10)

mtz = np.random.randint(1,50, [4, 4])

mediaColunas = mtz.mean(axis=0)
mediaLinhas = mtz.mean(axis=1)


print(f"Médias das colunas: {mediaColunas}")
print(f"Médias das linhas: {mediaLinhas}")

maiorValorColuna = mediaColunas.max()
maiorValorLinha = mediaLinhas.max()

print(f"Maior valor da média das colunas: {maiorValorColuna}")
print(f"Maior valor da média das linhas: {maiorValorLinha}")

print(np.unique(mtz, return_counts = True))

valores, contagens = np.unique(mtz, return_counts=True)

numeros2x = valores[contagens == 2]
print(numeros2x)
```
