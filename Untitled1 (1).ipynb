Exercício 7

```python
palavra = input("Entre com uma palavra: ")

tamanho = len(palavra)

contadorVogais = 0
letraA = False
vogais = 'AEIOU'

for i in range(tamanho):
  print(palavra[i].upper())

  if(palavra[i].upper() in vogais):
    contadorVogais+=1

  if(palavra[i].upper() == 'A'):
    letraA = True

print("Número Vogais:", contadorVogais)
print("Número de letras A: ", letraA)
```

```python
numero1 = int(input("Entra com o primeiro número: "))
numero2 = int(input("Entra com o segundo número: "))

print(numero1 + numero2)
if((numero1 + numero2) % 2 == 0):
  print("A Soma dos 2 números é par")
else:
  print("A Soma dos 2 números é ímpar")
print(numero1 - numero2)
print(numero1 * numero2)
print(numero1 / numero2)
print(numero1 % numero2)
print(numero1 ** numero2 )
```
