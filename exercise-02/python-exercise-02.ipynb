```python
nome1 = input("Digite seu nome: ")
peso1 = float(input("Digite seu peso: "))

nome2 = input("Digite seu nome: ")
peso2 = float(input("Digite seu peso: "))

nome3 = input("Digite seu nome: ")
peso3 = float(input("Digite seu peso: "))

nomes = (nome1, nome2, nome3)
pesos = (peso1, peso2, peso3)

maiorpeso = pesos.index(max(pesos))
menorpeso = pesos.index(min(pesos))

print(f"Pessoa mais pesada: {nomes[maiorpeso]}")
print(f"Pessoa mais leve: {nomes[menorpeso]}")


```

Exercício 8

```python
produtos = []

for i in range(3):
    nome = input("Digite o nome do produto: ")
    preco = float(input("Digite o preço do produto: "))
    quantidade = int(input("Digite a quantidade em estoque: "))

    dados = {}
    dados['nome'] = nome
    dados['preco'] = preco
    dados['quantidade'] = quantidade

    produtos.append(dados)

for produto in produtos:
    valorTotal = produto['preco'] * produto['quantidade']
    print(produto['nome'], valorTotal)
```
